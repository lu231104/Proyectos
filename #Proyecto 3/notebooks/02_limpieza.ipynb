```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/raw/bono_Independiente_trabajaperu.csv", encoding="latin-1")
```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.upper()
    .str.replace(" ", "_")
)
```

```python
columnas_texto = df.select_dtypes(include="object").columns

for columna in columnas_texto:
    df[columna] = df[columna].str.strip()
```

```python
df["MONTO"] = pd.to_numeric(df["MONTO"], errors="coerce")
```

```python
df["FECHA_COBRO"] = pd.to_datetime(
    df["FECHA_COBRO"],
    errors="coerce",
    utc=True
)
```

```python
df["FECHA_ACTUALIZACION"] = pd.to_datetime(
    df["FECHA_ACTUALIZACION"],
    errors="coerce",
    utc=True
)
```

```python
df["ANIO_COBRO"] = df["FECHA_COBRO"].dt.year
```

```python
df["MES_COBRO"] = df["FECHA_COBRO"].dt.month
```

```python
df["NOMBRE_MES"] = df["FECHA_COBRO"].dt.month_name()
```

```python
df["TRIMESTRE_COBRO"] = df["FECHA_COBRO"].dt.quarter
```

```python
df["PERIODO_COBRO"] = df["FECHA_COBRO"].dt.to_period("M").astype(str)
```

```python
df["BONO_COBRADO_FLAG"] = np.where(
    df["BONO_COBRADO"].eq("SI"),
    1,
    0
)
```

```python
df.duplicated().sum()
```

```python
df.to_csv(
    "../data/processed/bonos_clean.csv",
    index=False
)
```
